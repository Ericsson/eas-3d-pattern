# Deprecations and Migration

Every deprecated name in `eas_3d_pattern` raises a `FutureWarning` (visible by default) and
logs a warning on the `eas_3d_pattern` logger. This notebook lists them and shows, for each,
the old call that warns and the migration that does not.

## Deprecation summary

| # | Deprecated | Use instead |
|---|------------|-------------|
| 1 | `AntennaPattern.raw_data` | `AntennaPattern.data` |
| 2 | `AntennaPattern.Pattern_3D` | `AntennaPattern.pattern` |
| 3 | `AntennaPattern.is_nonuniform_sampling` | `not AntennaPattern.is_uniform_sampling` |
| 4 | `SectorDefinition(load_default=..., top_border=...)` | `sector.from_preset("eas", top_border=...)` / `Sector()` |
| 5 | `SectorDefinition.presets()` | `sector.preset_names()` |
| 6 | `SectorDefinition.validate_preset(name)` | `sector.validate_preset(name)` |
| 7 | `SectorDefinition.from_preset(name, ...)` | `sector.from_preset(name, ...)` |
| 8 | `BoundaryBoxSquare` | `BoundaryBox` |

All eight emit `FutureWarning`. The whole `SectorDefinition` class (rows 4-7) is a compatibility layer scheduled for removal, so every entry point into it warns.

## Setup

```python
import warnings

from eas_3d_pattern import SAMPLE_JSON, AntennaPattern
from eas_3d_pattern.sector import (
    BoundaryBox,
    BoundaryBoxSquare,
    Sector,
    SectorDefinition,
    from_preset,
    preset_names,
    validate_preset,
)

# `FutureWarning` is shown by default, but reported once per location.
# `simplefilter("always")` makes every cell below report on re-run.
warnings.simplefilter("always", FutureWarning)

pattern = AntennaPattern(SAMPLE_JSON[0])
TOP_BORDER = 85.0
```

## 1. `raw_data` -> `data`

The payload is normalized on load, so `raw_data` was a misnomer.

### Deprecated

```python
data = pattern.raw_data
print(f"type: {type(data).__name__}, len: {len(data)}")
```

### Migration

```python
data = pattern.data
print(f"type: {type(data).__name__}, len: {len(data)}")
```

## 2. `Pattern_3D` -> `pattern`

Renamed to the PEP 8-compliant `pattern`.

### Deprecated

```python
ds = pattern.Pattern_3D
print(f"type: {type(ds).__name__}, items: {list(ds.sizes)}")
```

### Migration

```python
ds = pattern.pattern
print(f"type: {type(ds).__name__}, items: {list(ds.sizes)}")
```

## 3. `is_nonuniform_sampling` -> `not is_uniform_sampling`

Redundant with negating the positive form.

### Deprecated

```python
print("non-uniform:", pattern.is_nonuniform_sampling)
```

### Migration

```python
print("non-uniform:", not pattern.is_uniform_sampling)
```

## 4. `SectorDefinition(...)` -> `from_preset(...)` / `Sector()`

The old constructor loaded the EAS preset by default, so `SectorDefinition()` was not empty.
The replacement splits the two intents: `from_preset` for a populated collection, `Sector()` for
an empty one.

### Deprecated

```python
preset_sectors = SectorDefinition(load_default=True, top_border=TOP_BORDER)
empty_sectors = SectorDefinition(load_default=False)
print("preset:", len(preset_sectors.sectors), "sectors | empty:", len(empty_sectors.sectors))
```

### Migration

```python
preset_sectors = from_preset("eas", top_border=TOP_BORDER)
empty_sectors = Sector()
print("preset:", len(preset_sectors.sectors), "sectors | empty:", len(empty_sectors.sectors))
```

## 5. `SectorDefinition.presets()` -> `preset_names()`

### Deprecated

```python
print(SectorDefinition.presets())
```

### Migration

```python
print(preset_names())
```

## 6. `SectorDefinition.validate_preset()` -> `validate_preset()`

### Deprecated

```python
SectorDefinition.validate_preset("eas")
print("valid")
```

### Migration

```python
validate_preset("eas")
print("valid")
```

## 7. `SectorDefinition.from_preset()` -> `from_preset()`

### Deprecated

```python
sectors = SectorDefinition.from_preset("eas", top_border=TOP_BORDER)
print(f"type: {type(sectors).__name__}, number of sectors: {len(sectors.sectors)}")
```

### Migration

```python
sectors = from_preset("eas", top_border=TOP_BORDER)
print(f"type: {type(sectors).__name__}, number of sectors: {len(sectors.sectors)}")
```

## 8. `BoundaryBoxSquare` -> `BoundaryBox`

The `"Square"` suffix was redundant. `BoundaryBoxSquare` is now a warning subclass of
`BoundaryBox`; `isinstance(x, BoundaryBox)` still holds, but the two never compare equal.

### Deprecated

```python
box = BoundaryBoxSquare((0.0, "<="), (10.0, "<"), (0.0, "<="), (20.0, "<"))
print(box, "| isinstance BoundaryBox:", isinstance(box, BoundaryBox))
```

### Migration

```python
box = BoundaryBox((0.0, "<="), (10.0, "<"), (0.0, "<="), (20.0, "<"))
print(box, "| isinstance BoundaryBox:", isinstance(box, BoundaryBox))
```
